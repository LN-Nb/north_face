Il n'y a aucun doublon exact, donc nous ne supprimerons pas de lignes.

```python
def text_cleaning(text): 
    """Mettre en minuscules,
    Remplacer la ponctuation et les caractères spéciaux par des espaces, Supprimer les espaces en trop"""

    text = text.lower()
    text = re.sub(r"[\W_]+", " ", text)
    return text.strip()

data["clean_text"] = data["raw"].fillna("").apply(text_cleaning)
```

```python
vectorizer = TfidfVectorizer(
    stop_words="english",
    max_df=0.95,
    min_df=2
)

# Remplace les valeurs manquantes par une chaîne vide
tfidf_matrix = vectorizer.fit_transform(data["clean_text"].fillna(""))

print("Dimensions :", tfidf_matrix.shape)
print("Token samples :", vectorizer.get_feature_names_out()[:20])
```

```python
pd.DataFrame(
    tfidf_matrix[:5].toarray(),
    columns=vectorizer.get_feature_names_out()
)
```

```python
dbscan = DBSCAN(
    eps=0.5,
    min_samples=10,
    metric="cosine"
)
```

```python
data["cluster"] = dbscan.fit_predict(tfidf_matrix)

display(data["cluster"].value_counts().sort_index())
```

```python
display(data.sort_values("cluster"))
```

```python
print(f"cluster_nb : {len(set(data['cluster'])) - 1}")  # On soustrait 1 pour exclure le bruit (-1)
```

Le cluster -1 est le cluster considéré comme du bruit par l'algorithme Dbscan. Nous allons tenter de réduire le nombre de produits dans ce cluster en ajustant les paramètres de l'algorithme.

```python
parametres = {
    "eps": [0.4, 0.5, 0.6, 0.7, 0.8, 0.9, 1.0],
    "min_samples": [2, 3, 4, 5, 6, 7, 8, 9, 10, 12, 15]
}

results = []

for params in ParameterGrid(parametres):
    clusters = DBSCAN(**params, metric="cosine").fit_predict(tfidf_matrix)
    without_noise = clusters != -1
    try : 
        score = silhouette_score(
            tfidf_matrix[without_noise],
            clusters[without_noise],
            metric="cosine"
        )
    except ValueError:
        score = np.nan

    results.append({
        **params,
        "silhouette": score,
        "noise_pct": (1 - without_noise.mean()) * 100,
        "cluster_nb": len(set(clusters)) - (1 if -1 in clusters else 0)
    })

results = pd.DataFrame(results)
display(results.sort_values("cluster_nb", ascending=False))
```

```python
candidates = results[(results["cluster_nb"] <= 20) & (results["cluster_nb"] >= 10) & results["silhouette"].notna()]
```

```python
display(candidates)
```

```python
if candidates.empty:
    print("No model valid under 20 clusters.")
else:
    best = candidates.sort_values("silhouette", ascending=False).iloc[0]
    display(best)
```

```python
import matplotlib.pyplot as plt

plt.scatter(results["noise_pct"], results["silhouette"])
plt.xlabel("Noise proportion")
plt.ylabel("Silhouette's score")
plt.show()
```

Ce modele montre un score de silhouette difficile à choisir puisque dans les clusters selctionnés, le score de silhouette est très faible. Nous allons travailler sur la qualité des données pour améliorer le score de silhouette.
